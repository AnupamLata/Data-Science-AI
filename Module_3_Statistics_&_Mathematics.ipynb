**Descriptive Statistics** :

Descriptive statistics is the branch of statistics that summarizes , organizes, and describes a dataset using numerical values and visualizations.

**Types of Descriptive Statistics:**

  * Measures of central tendency
  * Measure of dispersion
  * Measure of distribution

**1. Measure of Central tendency**

Measure of Central Tendancy help us find the "middle" or "typical" value of a collection od data.

**1. Mean**

Average value of all values Formula Mean = ∑x/n
1. ∑x = Sum of all observation
2. n= Number of observation
