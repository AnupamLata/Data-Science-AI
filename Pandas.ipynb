#1.Creating series
A series is a one-dimensional labeled array tha stores a single column of data.

```python
import pandas as pd
marks = pd.Series([85,90,78,92,88])

print(marks)
```

Creating a series of strings

```python
import pandas as pd

names = pd.Series(["Anupam", "Anushka", "Aniket", "Anshu"])

print(names)
```

Creating a series with custom index

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 78],
    index=["Maths","Science","English"]
)

print(marks)
```

Create a Series From a Dictionary

```python
import pandas as pd

student = pd.Series({
    "Name": "Anupam",
    "age" : 22,

})
```
